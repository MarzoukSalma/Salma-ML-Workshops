### C'est quoi le Reinforcement Learning exactement ?

Le **Reinforcement Learning (RL)** est une façon d'apprendre inspirée de la psychologie comportementale.
L'idée est simple :
si tu fais quelque chose de bien → tu reçois une récompense. 

Si tu fais quelque chose de mal → tu reçois une pénalité. 

Avec le temps, tu apprends à faire les bonnes actions.

### Définitions fondamentales
Tout le RL tourne autour de **6 concepts clés**. Une fois qu'on les maîtrise, tout le reste devient logique.
1. **L'Agent** :
C'est l'entité qui apprend et prend des décisions. C'est le "cerveau" du système.
> Exemples : un robot, un programme qui joue aux échecs, une voiture autonome

2. **L'Environnement (Environment)** :
C'est tout ce avec quoi l'agent interagit. L'agent ne le contrôle pas directement — il peut juste l'observer et agir dessus.
> Exemples : un labyrinthe, un jeu vidéo, la route pour une voiture autonome

3. **L'État (State — s)** :
C'est la description de la situation actuelle de l'environnement. Ce que l'agent "voit" à un instant t.
> Exemple dans un labyrinthe : la position actuelle de l'agent (ligne 2, colonne 3)

4. **L'Action (a)** :
C'est ce que l'agent peut faire dans un état donné. L'ensemble de toutes les actions possibles s'appelle l'espace d'actions.

>Exemple dans un labyrinthe : aller à gauche, droite, haut, bas

5. **La Récompense (Reward — r)** :
C'est un signal numérique que l'environnement envoie à l'agent après chaque action. Elle guide l'apprentissage.




6. **La Politique (Policy — π)** :
C'est la stratégie de l'agent — une fonction qui dit quoi faire dans chaque état. C'est ce qu'on cherche à optimiser.
>Au début : politique aléatoire (l'agent fait n'importe quoi)

>À la fin : politique optimale (l'agent sait exactement quoi faire)


### Comment tout ça s'articule ?

## Les familles d'algorithmes RL

Il existe plusieurs façons pour un agent d'apprendre. On les regroupe en 3 grandes familles :


### 1. Value-Based = “je calcule avant d’agir”

Ici, l’agent se pose toujours la question :
> Quelle action me donne le meilleur score ?

Il construit une table (Q-table) comme ça



Il choisit l’action avec la plus grande valeur (ici : Droite)

**Algorithmes** :Q-Learning, DQN (version deep learning)

>Limite : 
Devient difficile si l’environnement est très grand (trop de valeurs à stocker)

### 2.Policy-Based 

Le Policy-Based Reinforcement Learning consiste à apprendre directement une politique (policy) :

> une fonction qui associe chaque état à une action (ou une probabilité d’action)

On note généralement : $\pi(a \mid s)$

➡️ Probabilité de choisir l’action a dans l’état s

**Fonctionnement**:
1. L’agent observe un état s
2. La policy π(a∣s) donne une distribution de probabilités
3. Une action est choisie
4. L’agent reçoit une récompense
5. La policy est mise à jour pour favoriser les bonnes actions

La policy est souvent modélisée par : un réseau de neurones paramétré πθ	​

Entrée : état

Sortie : probabilités des actions

**Objectif**: Maximiser l’espérance de la récompense cumulée 

$J(\theta) = \mathbb{E}_{\pi_\theta}[R]$

où :

θ = paramètres du modèle

R = récompense cumulée

**Mise à jour**:

Les paramètres θ sont ajustés pour :

augmenter la probabilité des actions qui donnent une bonne récompense
diminuer celles qui donnent une mauvaise récompense

c’est le principe du Policy Gradient

**Algorithmes**: REINFORCE (simple, mais instable)
PPO (Proximal Policy Optimization)

👉 utilisé dans ChatGPT




### 3. Model-Based Reinforcement Learning
Le Model-Based Reinforcement Learning consiste à :
> apprendre un modèle de l’environnement + utiliser ce modèle pour planifier les actions

**Les 2 composants essentiels**:

1️⃣ Le modèle de l’environnement : C’est une fonction qui apprend 

\begin{equation*}
(s, a) \rightarrow (s', r)
\end{equation*}

c’est-à-dire :

• état actuel : s

• action : a

• prochain état : s′

• récompense : r

L’agent apprend :

> si je fais ça → voilà ce qui va arriver

2️⃣ Le module de planification

Une fois le modèle appris :

> l’agent peut simuler le futur sans agir réellement

3. **Cycle complet d’apprentissage**:
*Étapes* :

•Observer l’état s

•Choisir une action a

•Observer s′,r

•Apprendre le modèle

•Simuler plusieurs actions possibles

•Choisir la meilleure

•Répéter


| Type            |  Comment il pense                 |  Comment il choisit                     |  Est-ce qu’il réfléchit au futur ? |
| --------------- | ----------------------------------- | ----------------------------------------- | ------------------------------------ |
|  Value-Based  | Compare les scores de chaque action | Prend l’action avec le meilleur score     |  Non                                |
|  Policy-Based | Apprend directement quoi faire      | Donne directement une action              | Non                                |
|  Model-Based | Imagine ce qui va se passer         | Teste plusieurs possibilités puis choisit | Oui                                |
